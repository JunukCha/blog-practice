### Minimal 스타터 생성·설치·개발 서버

```python
%%bash
git config --global user.name tester
git config --global user.email t@example.com
curl -fLSs https://raw.githubusercontent.com/github/replica/main/scripts/install-starter.sh -o install-starter.sh
bash install-starter.sh --list
bash install-starter.sh --starter minimal my-prototype
cd my-prototype
npm install
cat package.json | grep -A12 '"scripts"'
(nohup npm run dev -- --port 5173 --host 127.0.0.1 > dev.log 2>&1 &)
sleep 5
cat dev.log
curl -s http://127.0.0.1:5173/ | head -20
```
