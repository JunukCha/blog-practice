### Claude Code 2.1.327 설치

```python
%%bash
set -eo pipefail
mkdir -p /content/cc && cd /content/cc
npm init -y >/dev/null
npm install @anthropic-ai/claude-code@2.1.327 2>&1 | tail -5
```

### 버전 확인과 claude update

```python
%%bash
cd /content/cc
export HOME=/tmp/h; mkdir -p $HOME
npx claude --version
timeout 60 npx claude update 2>&1 | tail -5
```

### 불완전한 규칙 시작 경고 확인

```python
%%bash
cd /content/cc
export HOME=/tmp/h
timeout 60 npx claude -p hi --allowedTools "Bash(git" 2>&1 | head -10
echo ----
timeout 60 npx claude -p hi --disallowedTools "Bash(git" --tools "Bash(git" 2>&1 | head -10
```
