# Claude-BugHunter 정리: 버그 바운티용 Claude Code 스킬 85개·명령어 16개를 플러그인으로 설치하는 법, 쓸 수 있는 범위와 Anthropic 보안 제한

위에서부터 차례로 실행해 보세요.

### 플러그인 설치

```python
%%bash
npm i -g @anthropic-ai/claude-code
claude plugin marketplace add elementalsouls/Claude-BugHunter
claude plugin install claude-bughunter@elementalsouls
claude plugin list
```

### 스킬·명령어 개수 세기

```python
%%bash
P=$HOME/.claude/plugins/cache/elementalsouls/claude-bughunter/2.1.0
ls $P/skills | wc -l
ls $P/skills | grep -c '^hunt-'
ls $P/commands | wc -l
ls $P/commands
```

### cbh CLI 설치

```python
%%bash
apt-get update && apt-get install -y pipx
pipx install git+https://github.com/elementalsouls/Claude-BugHunter
~/.local/bin/cbh --help
```

### Juice Shop 실행

```python
%%bash
mkdir -p /content/js && cd /content/js
curl -sL -o js.tgz https://github.com/juice-shop/juice-shop/releases/download/v20.4.0/juice-shop-20.4.0_node22_linux_x64.tgz
tar xzf js.tgz && cd juice-shop*/ && (nohup npm start > /content/js.log 2>&1 &)
sleep 30; curl -sI http://localhost:3000 | head -1
```

### 스코프 검증과 분류

```python
%%bash
P=$HOME/.claude/plugins/cache/elementalsouls/claude-bughunter/2.1.0
python3 $P/engine/scope.py localhost 127.0.0.1 example.com --in-scope localhost --in-scope 127.0.0.1; echo exit=$?
~/.local/bin/cbh classify 'http://localhost:3000/rest/products/search?q=apple'
```
