# OpenAI가 따르겠다고 한 robots.txt 'AI 학습 거부' 공통 신호 정리: 기존 GPTBot 차단과 무엇이 다른지, 내 블로그 robots.txt에 어떻게 적는지

위에서부터 차례로 실행해 보세요.

### 토큰 그룹 판정 및 Allow/Disallow 순서 비교

```python
%%bash
cat > t2.py <<'E'
from urllib.robotparser import RobotFileParser
for order in (["Disallow: /","Allow: /press-kit/"],["Allow: /press-kit/","Disallow: /"]):
    r=RobotFileParser(); r.parse(["User-agent: __usage_ai-training"]+order)
    print(order,[(x,r.can_fetch("__usage_ai-training",x)) for x in ["/","/blog/a","/press-kit/logo.png"]])
E
python3 t2.py
```

### * 규칙만 있을 때 토큰 판정

```python
%%bash
cat > t.py <<'E'
from urllib.robotparser import RobotFileParser
B="User-agent: *\nDisallow: /private/\n"
C="User-agent: *\nDisallow: /\n"
for n,t in [("B",B),("C",C)]:
    r=RobotFileParser(); r.parse(t.splitlines())
    for ag in ["__usage_ai-training","GPTBot"]:
        print(n,ag,[(x,r.can_fetch(ag,x)) for x in ["/","/blog/post1","/press-kit/logo.png","/private/a"]])
E
python3 t.py
```
