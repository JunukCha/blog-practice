# Gemini 2.5 Pro·Flash 12월 1일 자동 전환 대비: 어떤 모델로 바뀌고 요금은 어떻게 되는지, thinking_budget→thinking_level 등 깨지는 변경 체크리스트

위에서부터 차례로 실행해 보세요.

### 예제 프로젝트 만들기

```python
%%bash
mkdir -p proj && cd proj
cat > app.py <<'E'
from google import genai
from google.genai import types
client = genai.Client(api_key="YOUR_API_KEY")
resp = client.models.generate_content(
    model="gemini-2.5-pro",
    contents="Explain RAG in 3 lines",
    config=types.GenerateContentConfig(
        temperature=0.7, top_p=0.9, top_k=40, candidate_count=1,
        thinking_config=types.ThinkingConfig(thinking_budget=2048)))
print(resp.text)
E
cat > summarize.py <<'E'
MODEL = "gemini-2.5-flash"
cfg = {"temperature": 0.2, "thinking_config": {"thinking_budget": 0}}
E
```

### grep으로 고칠 곳 찾기

```python
%%bash
cd proj && grep -rnE "gemini-2\.5-(pro|flash)|thinking_budget|temperature|top_p|top_k|candidate_count" .
```
