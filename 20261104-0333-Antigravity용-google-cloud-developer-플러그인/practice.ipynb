# Antigravity용 google-cloud-developer 플러그인 사용법: 최신 공식 문서를 근거로 답하게 하는 MCP와 과금 API·삭제 명령을 막는 가드레일이 에이전트 작업을 어떻게 바꾸는지

위에서부터 차례로 실행해 보세요.

### Antigravity CLI 설치

```python
%%bash
curl -fsSL https://antigravity.google/cli/install.sh | bash
export PATH="$HOME/.local/bin:$PATH"
agy --version
```

### 플러그인 설치와 설치된 구성 확인

```python
%%bash
export PATH="$HOME/.local/bin:$PATH"
agy plugin install https://github.com/google/skills/plugins/cloud/google-cloud-developer
agy plugin list
find ~/.gemini/config/plugins/google-cloud-developer -type f | sort
cat ~/.gemini/config/plugins/google-cloud-developer/mcp_config.json
cat ~/.gemini/config/plugins/google-cloud-developer/rules/google-cloud-discovery.md
```

### gcloud SKILL.md 원문에서 가드레일 규칙 읽기

```python
%%bash
git clone --depth 1 https://github.com/google/skills.git
cd skills/skills/cloud/gcloud
grep -n -i -E 'MUST|Proactive|delete|billing|Denylist' SKILL.md
sed -n 195,235p SKILL.md
```
