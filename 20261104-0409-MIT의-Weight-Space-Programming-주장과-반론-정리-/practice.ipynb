### 프롬프트 4개 받기, 가중치 외 diff, sha256 확인

```python
%%bash
set -eo pipefail
mkdir -p prompts && cd prompts
for g in breakout pong pacman freeway; do mkdir -p $g; curl -sfL -o $g/prompt.txt https://raw.githubusercontent.com/RikiyaT/lm_understand-logs/refs/heads/main/exp1_which_game/runs/$g/prompt.txt; done
wc -c */prompt.txt
sha256sum */prompt.txt
for g in pong pacman freeway; do echo "== breakout vs $g (소수점 숫자 제거 후 diff)"; diff <(sed -E 's/-?[0-9]+\.[0-9]+//g' breakout/prompt.txt) <(sed -E 's/-?[0-9]+\.[0-9]+//g' $g/prompt.txt) && echo '차이 없음'; done
```
